```python
count = 0
while count < 5:
  print(f"{count} is less than 5")
  count = count + 1
else:
    print(f"{count} is not less than 5")
```

```python
count = 0
while (count == 0):
  count += 1
  print(count)
  break
else:
  print("No Break")
```
